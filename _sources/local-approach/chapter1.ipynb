### Running models locally
Google Colab provides some access to GPUs for free, although there are limits on usage. Sometimes they'll stop your code while it's running. Other times, they'll say that you've exceeded your quota. We don't want to run into those issues during the workshop. We'll provide code to use Colab if you want and the endpoint will still be there as a fallback.You can also pay for a Google Colab Pro account that gives you more access to GPUs, but we're not here to recommend or sell you a service.

If you have a GPU or Apple Silicon chip (M1/M2), you can also run the model locally on your computer. Simply download the notebook from Colab and run it in Jupyter Notebook or VS Code. Please only follow this approach if you have worked with models before and are comfortable installing packages and troubleshooting issues. We recommend the endpoint approach for most participants, but want to provide this option for those who may prefer to manage their own environment.
