# Setting Up Your own Model Endpoint


For example, if you'd like to use Alibaba's servers, you just need to change the base_url and enter an API key. Here are [instructions](https://www.alibabacloud.com/help/en/model-studio/get-api-key) on how to get an Alibaba key. There are many endpoint providers. [OpenRouter](https://openrouter.ai/qwen/qwen3-vl-8b-instruct) is a useful directory with cost comparisons. It is, of course, also possible to run models on your own hardware.    

```python 
base_url="https://dashscope-intl.aliyuncs.com/compatible-mode/v1"
api_key="sk-..."
model="qwen2.5-vl-7b-instruct"
```
